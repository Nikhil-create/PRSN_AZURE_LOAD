```python
catalog_name = spark.sql("SELECT current_catalog()").collect()[0][0]
```

```python
%run ./utils
```

```python
df = (spark.read
      .format("parquet")
      .option("header", "true")
      .load("/Volumes/one_db/raw/raw/customer_info/")
)
```

```python
if not_exists(f"/Volumes/{catalog_name}/raw/stage/customer_info_load"):
    df.write.mode("overwrite").format("parquet").save(f"/Volumes/{catalog_name}/raw/load/customer_info/")
    dbutils.fs.put(f"/Volumes/{catalog_name}/raw/stage/customer_info_load","",overwrite=True)
    print("load completed")
else:
    print("load skipped")
```

```python
if not_exists(f"/Volumes/{catalog_name}/raw/stage/customer_info_delta"):
    df.write.mode("overwrite").format("delta").save(f"abfss://raw@oneadls.dfs.core.windows.net/data/delta/customer_info/")
    dbutils.fs.put(f"/Volumes/{catalog_name}/raw/stage/customer_info_delta","",overwrite=True)
    print("load completed")
else:
    print("load skipped")
```

```python
stage_path = f"/Volumes/{catalog_name}/raw/stage/"
for file in dbutils.fs.ls(stage_path):
    dbutils.fs.rm(file.path, recurse = True)
print("files deleted")
```
